## 7. Handle Missing Values

Missing Age values are replaced with the median age.

Missing Embarked values are replaced with the most frequently occurring value.

The Cabin column is removed because it contains a very large proportion of missing values.

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df = df.drop(columns=['Cabin'])
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```
